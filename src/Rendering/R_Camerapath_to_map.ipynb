```python
import pandas as pd
import matplotlib.pyplot as plt
import tkinter as tk
from tkinter import filedialog
import os

# Pick CSV file
root = tk.Tk()
root.withdraw()
```

```python
csv_path = filedialog.askopenfilename(title='Select CSV', filetypes=[('CSV files', '*.csv')])
print(f'Selected: {csv_path}')
```

```python
# Load CSV and extract columns B (index 1) and C (index 2)
df = pd.read_csv(csv_path, header=0)
x = df.iloc[:, 1].values  # column B
y = df.iloc[:, 2].values  # column C
print(f'Loaded {len(x)} points')
print(df.iloc[:5, [1, 2]])
```

```python
fig, ax = plt.subplots(figsize=(10, 8))

fig.patch.set_alpha(0)
ax.patch.set_alpha(0)

ax.plot(x, y, color='red', linewidth=1.5)
ax.set_aspect('equal')  # preserve true shape of path
ax.axis('off')

plt.tight_layout()
folder = '03_camerapath_GFX'
filename = '01_walkb.png'
filename = filename
savepath = os.path.join(folder, filename)
plt.savefig(savepath, transparent=True, bbox_inches='tight', dpi=150)
plt.show()
print('Saved to snaking_path.png')
```

```python
# Plot snaking path: red line on completely transparent background
fig, ax = plt.subplots(figsize=(10, 8))

# Transparent backgrounds
fig.patch.set_alpha(0)
ax.patch.set_alpha(0)

# Red snaking path
ax.plot(x, y, color='red', linewidth=1.5)

# Remove all axes decoration
ax.axis('off')

plt.tight_layout()
plt.savefig('snaking_path.png', transparent=True, bbox_inches='tight', dpi=150)
plt.show()
print('Saved to snaking_path.png')
```
